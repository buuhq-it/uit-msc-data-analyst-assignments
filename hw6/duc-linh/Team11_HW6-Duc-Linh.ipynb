```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("input.csv")
df.head()
import numpy as np

# Prepare features and labels
X = df[['income', 'amount', 'down_payment']].values
y = df['label_refused'].values

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Initialize parameters
weights = np.zeros(X.shape[1])
bias = 0
learning_rate = 0.01
steps = 3

# Track weights after each step
history = []

# Perform gradient descent
for step in range(steps):
    z = np.dot(X, weights) + bias
    predictions = sigmoid(z)

    # Compute gradients
    error = predictions - y
    dw = np.dot(X.T, error) / len(X)
    db = np.sum(error) / len(X)

    # Update parameters
    weights -= learning_rate * dw
    bias -= learning_rate * db

    # Store current state
    history.append({
        'step': step + 1,
        'weights': weights.copy(),
        'bias': bias
    })

# Convert history to DataFrame for display
history_df = pd.DataFrame(history)

```

```python
# New data
new_data = np.array([
    [-0.746314279, -1.570549473, -1.898252746],
    [-0.746314279,  1.392803082, -0.983993986],
    [ 0.727299903, -1.526046325, -1.893831903],
    [ 0.204404548, -1.305812309, -1.863408901],
    [ 0.204404548, -1.607190177, -1.868162495],
    [ 0.204404548, -1.411864994, -1.830133742],
    [-0.033275159, -1.174945862, -1.858702843],
    [-0.651242397, -0.793850221, -1.820626554],
    [ 0.204404548, -1.279239718, -1.697033106],
    [ 0.442084255, -1.700978589, -1.887176872]
])

# Use final weights and bias from previous training
final_weights = history[-1]['weights']
final_bias = history[-1]['bias']

# Compute predictions
z_new = np.dot(new_data, final_weights) + final_bias
pred_probs = sigmoid(z_new)
pred_labels = (pred_probs >= 0.5).astype(int)

# Create output DataFrame
predictions_df = pd.DataFrame(new_data, columns=['income', 'amount', 'down_payment'])
predictions_df['predicted_probability'] = pred_probs
predictions_df['predicted_label'] = pred_labels
predictions_df
```
