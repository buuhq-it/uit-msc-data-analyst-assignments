```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
p = np.linspace(0.01, 0.99, 200)
logit = np.log(p / (1 - p))

plt.figure(figsize=(8, 4))
plt.plot(p, logit, color='green', label='Hàm Logit')
plt.axhline(y=0, color='gray', linestyle='--')
plt.axvline(x=0.5, color='gray', linestyle='--')
plt.xlabel('p (Xác suất)')
plt.ylabel('logit(p)')
plt.title('Đồ thị hàm Logit')
plt.grid(True)
plt.legend()
plt.show()
```

```python
z = np.linspace(-10, 10, 200)
sigma = 1 / (1 + np.exp(-z))

plt.figure(figsize=(8, 4))
plt.plot(z, sigma, label='Hàm Sigmoid')
plt.axhline(y=0.5, color='gray', linestyle='--')
plt.axvline(x=0, color='gray', linestyle='--')
plt.xlabel('z')
plt.ylabel('σ(z)')
plt.title('Đồ thị hàm Sigmoid')
plt.grid(True)
plt.legend()
plt.show()
```
