```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import time
from io import StringIO
```

```python
cleaned_pages = []
max_pages = 14  # Tùy chỉnh theo nhu cầu (~20 dòng/trang)

for page in range(1, max_pages + 1):
    url = f"https://www.cophieu68.vn/quote/history.php?cP={page}&id=vnm"
    resp = requests.get(url)
    resp.encoding = 'windows-1258'
    soup = BeautifulSoup(resp.text, 'html.parser')

    table = soup.find("table", {"id": "history"})
    if not table:
        print(f"⚠️ Không tìm thấy bảng ở trang {page}")
        continue

    try:
        df = pd.read_html(str(table))[0]

        # ❌ Bỏ dòng trống / dòng tiêu đề lỗi / dòng rác
        df = df[df.notna().sum(axis=1) >= 2]
        df = df[~df.iloc[:, 0].astype(str).str.strip().eq('')]
        df = df[~df.iloc[:, 0].astype(str).str.contains("Ngày|GiÃ¡|NN mua|NN bÃ¡n", na=False)]

        # ❌ Nếu dòng đầu tiên vẫn là tiêu đề -> loại bỏ
        if df.shape[0] > 0 and df.iloc[0, 0].strip().lower() in ['ngày', 'ngay', 'date']:
            df = df.iloc[1:]

        # ❌ Nếu có cột rác -> chỉ giữ 9 cột đầu tiên
        if df.shape[1] > 9:
            df = df.iloc[:, :9]

        cleaned_pages.append(df)
        print(f"✅ Tải và lọc xong trang {page}")

    except Exception as e:
        print(f"❌ Lỗi xử lý trang {page}: {e}")

    time.sleep(1)

# 🧩 Gộp các trang đã được làm sạch
df = pd.concat(cleaned_pages, ignore_index=True)
```

```python
# 🏷️ Đặt lại tên cột tiếng Anh
df.columns = [
    'Date', 'Matched Price', 'Volume', 'Open Price',
    'High Price', 'Low Price', 'Foreign Buy',
    'Foreign Sell', 'Foreign Value'
]
# 📆 Chuyển ngày
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y', errors='coerce')
df = df.dropna(subset=['Date'])

# 📊 Sắp xếp
df = df.sort_values('Date').reset_index(drop=True)
# df = df.sort_values('Date', ascending=False).reset_index(drop=True)

df.to_csv("vinamilk_clean.csv", index=False)
```
