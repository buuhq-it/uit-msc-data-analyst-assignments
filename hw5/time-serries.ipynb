```python
# Data mẫu
data = [2.5, 3.6, 3.0, 5.1, 4.6, 6.2, 5.8, 6.9, 6.1, 7.5]

# Hàm tính residuals theo 1 cặp phi1 và theta1
def calculate_residuals(data, phi1, theta1):
    n = len(data)
    residuals = [0]  # epsilon_0 = 0
    for t in range(1, n):
        x_t = data[t]
        x_t_minus_1 = data[t-1]
        epsilon_prev = residuals[-1]
        epsilon_t = x_t - phi1 * x_t_minus_1 - theta1 * epsilon_prev
        residuals.append(epsilon_t)
    return residuals

# Hàm tính log-likelihood
def log_likelihood(residuals):
    n = len(residuals) - 1  # bỏ residual đầu tiên (do epsilon_0 giả định)
    sigma2 = sum(e**2 for e in residuals[1:]) / n
    logL = -n/2 * (np.log(2*np.pi) + np.log(sigma2) + 1)
    return logL

# Tìm kiếm phi1 và theta1 tốt nhất
import numpy as np

best_phi1 = None
best_theta1 = None
best_logL = -float('inf')

phi1_candidates = np.arange(-1.0, 1.05, 0.05)  # từ -1.0 tới 1.0 bước 0.05
theta1_candidates = np.arange(-1.0, 1.05, 0.05)

for phi1 in phi1_candidates:
    for theta1 in theta1_candidates:
        residuals = calculate_residuals(data, phi1, theta1)
        logL = log_likelihood(residuals)
        if logL > best_logL:
            best_logL = logL
            best_phi1 = phi1
            best_theta1 = theta1

print(f"Best phi1: {best_phi1:.4f}")
print(f"Best theta1: {best_theta1:.4f}")
print(f"Best log-likelihood: {best_logL:.4f}")
```
