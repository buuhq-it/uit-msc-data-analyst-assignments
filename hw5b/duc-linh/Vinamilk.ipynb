```python
import pandas as pd 

df = pd.read_csv(r'C:\Users\ducli\Downloads\vinamilk_stock.csv')
```

```python
# ADF test
from statsmodels.tsa.stattools import adfuller
adfuller(df['Matched Price'])
```

```python
# Sai phân bậc 1 
df['diff'] = df['Matched Price'].diff()
# ADF test
adfuller(df['diff'].dropna())
```

```python
df = df['diff'].dropna()
```

```python
# acf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt
plot_acf(df)
plot_pacf(df)
plt.show()
```
