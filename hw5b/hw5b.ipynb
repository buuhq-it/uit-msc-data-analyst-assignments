```python
import pandas as pd
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
import pmdarima
```

```python
df = pd.read_excel("hw5b-excel.xlsx", sheet_name="data", skiprows=3, usecols="C")
df.columns = ['Price']
# np.set_printoptions(precision=5)
# print(df['Price'].tail(10).values)
# df
```

```python
# arima_fit = pmdarima.auto_arima(df['Price'], error_action='ignore', suppress_warnings=True, seasonal=False)
# arima_fit.summary()
model = pmdarima.ARIMA(order=(1, 1, 1))
model.fit(df['Price'])  # y = chuỗi thời gian (list hoặc pd.Series)

# Dự báo n bước
forecast = model.predict(n_periods=4)
print("Dự báo 4 ngày tiếp theo (ARIMA(1,1,1)):")
for i, value in enumerate(forecast, 1):
    print(f"Day {i}: {value:.2f}")
```

```python
# Bước 2: Tạo mô hình ARIMA(1,1,1)
# model = ARIMA(df['Price'], order=(1, 1, 1), trend ='t') 

model = ARIMA(df['Price'], order=(1, 1, 1))
model_fit = model.fit()
print(model_fit.summary())
```

```python
# Bước 3: Dự báo 4 bước tiếp theo
forecast = model_fit.forecast(steps=4)
```

```python
# Bước 4: In kết quả
print("Dự báo 4 ngày tiếp theo (ARIMA(1,1,1)):")
for i, value in enumerate(forecast, 1):
    print(f"Day {i}: {value:.2f}")
```
